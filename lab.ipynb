### R_GLAB_386.3.9 - Adding Columns to DataFrames

```python
import pandas as pd 
import numpy as np
```

```python
data = {
    'Name': ['Jane', 'Princi', 'James', 'Fadi'],
    'Height': [5.1, 6.2, 5.1, 5.2],
    'Qualifications': ['Msc', 'MA', 'Msc', 'Msc'],
    'Score 1': [50, 86, 77, 45],
    'Score 2': [50, 96, 60, 30]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
address = ['NYC', 'NJ', 'PA', 'CA']

df['Address'] = address

print(df)
```

```python
# Create Total Score column
df['Total Score'] = df['Score 1'] + df['Score 2']

print(df)
```

```python
# Average Column
df['Average'] = df['Total Score'] / 2

print(df)
```

```python
# creating and initializing a list
values = [
    ['Rohan', 455],
    ['Elvish', 250],
    ['John', 495],
    ['Sai', 400],
    ['Eric', 350],
    ['Adam', 450]
]


columns = ['Name', 'Univ_marks']

# Cols: Name, Univ_marks

df = pd.DataFrame(values, columns=columns)
print(df)
```

```python
# Total Possible point is 500, we want the percentage to be a whole number above 1
# Create a column for grade percentage (total earned / total points possible) * 100

df['Grade_Percent'] = (df.Univ_marks/500) * 100

print(df)
```

```python
age = [20, 21, 19, 25, 33, 99]

df.insert(1, 'Age', age)

print(df)
```

```python
major = major = ['Art', 'Language', 'Math', 'Science', 'Nursing', 'Business']

df.insert(2, 'Major', major)

print(df)
```

```python
address = ['MI', 'GA', 'FL', 'NH', 'RI', 'MA']

# Assign does nOT change original DF in place, it returns a NEW dataframe
df1 = df.assign(address=address)
```
